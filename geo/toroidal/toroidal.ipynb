```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from ripser import ripser
from persim import plot_diagrams
```

```python
rng = np.random.default_rng()
u   = rng.uniform(0, 2*np.pi, 500) #phi
v   = rng.uniform(0, 2*np.pi, 500) #theta

# u, v = np.meshgrid(u, v)

def convert_angles_to_torus_xyz(u, v, R_major=1, r_tube=0.2):
    """Convert rad angles to torus coords"""
    x = np.cos(u) * (R_major + r_tube * np.cos(v))
    y = np.sin(u) * (R_major + r_tube * np.cos(v))
    z = r_tube * np.sin(v)
    return x, y, z

def convert_angles_to_sphere_xyz(u, v, R_major=1):
    """Convert rad angles to sphere coords"""
    x = R_major * np.cos(u) * np.sin(v)
    y = R_major * np.sin(u) * np.sin(v)
    z = R_major * np.cos(v)
    return x, y, z

def convert_xyz_to_angles(x, y, z):
    """Convert torus coords to rad angles"""
    u = np.arctan2(z/(x**2+y**2))
    v = np.arctan2(y/x)
    return u, v

R_major = 1
r_tube  = R_major/5

x, y, z = convert_angles_to_torus_xyz(u, v, R_major, r_tube)
# x, y, z = convert_angles_to_sphere_xyz(u, v, R_major)
geo_coordinates = np.column_stack((x, y, z))  # shape (N, 3)

```

```python
fig = plt.figure(figsize=(8, 12))
ax  = fig.add_subplot(projection='3d')
ax.scatter(x, y, z, color='red', alpha=0.7, s=10)

# force equal scaling
max_range = np.array([x.max()-x.min(), y.max()-y.min(), z.max()-z.min()]).max() / 2.0
mid_x = (x.max()+x.min()) * 0.5
mid_y = (y.max()+y.min()) * 0.5
mid_z = (z.max()+z.min()) * 0.5

ax.set_xlim(mid_x - max_range, mid_x + max_range)
ax.set_ylim(mid_y - max_range, mid_y + max_range)
ax.set_zlim(mid_z - max_range, mid_z + max_range)

plt.margins(0)
plt.show()
```
